# Data Understanding

### Dataset Columns Description



| Column Name | Description | 
|-------------|-------------|
| `index` | Sequential row identifier for each record in the dataset. | 
| `transaction_id` | Unique identifier assigned to each sales transaction. |  
| `transaction_date` | Date when the transaction occurred. |  
| `transaction_time` | Exact time when the transaction was processed. |  
| `store_id` | Unique identifier for the coffee shop branch/store. |  
| `store_location` | Physical location or branch name where the sale occurred. |  
| `product_id` | Unique identifier for the sold product. | 
| `transaction_qty` | Quantity of product units purchased in the transaction. |  
| `unit_price` | Price per single unit of the product. |  
| `Total_Bill` | Total amount paid for the transaction (`transaction_qty × unit_price`). |  
| `product_category` | Broad category of the sold product. |  
| `product_type` | Specific type or subcategory of the product. |  
| `product_detail` | Detailed product name or flavor. |  
| `Size` | Product serving size. | String | Large |
| `Month Name` | Full month name extracted from transaction date for easier analysis. |  
| `Day Name` | Day of the week extracted from transaction date. |  
| `Hour` | Hour of transaction extracted from transaction time. |  
| `Month` | Numeric month extracted from transaction date. | 
| `Day of Week` | Numeric representation of weekday (`0=Monday`, `6=Sunday`). | 

### Notes:
- **Primary business metrics:** `transaction_qty`, `unit_price`, `Total_Bill`
- **Time-based analysis columns:** `transaction_date`, `transaction_time`, `Month Name`, `Day Name`, `Hour`
- **Product analysis columns:** `product_category`, `product_type`, `product_detail`, `Size`
- **Store performance columns:** `store_id`, `store_location`

# EDA & Visualisation

```python
df.head(1)
```

#### What is the total revenue per day of the week?

```python
df_revperday = df.groupby(['Day of Week','Day Name'])['Total_Bill'].sum().reset_index().drop(columns= 'Day of Week').round(2)
df_revperday
```

```python
px.bar(data_frame=df_revperday, x='Day Name', y='Total_Bill', title='Total revenue per weekday', text_auto=True,
       labels={'Total_Bill': 'Total Revenue'}).show()
```

```python
# To find the day with the highest revenue
df_revperday.sort_values(by= 'Total_Bill', ascending= False).head(1)
```

####  Which months generate the highest revenue?

```python
df_revpermonth = df.groupby(['Month','Month Name'])['Total_Bill'].sum().reset_index().drop(columns= 'Month')
df_revpermonth
```

```python
px.bar(data_frame=df_revpermonth, x='Month Name', y='Total_Bill', title='Total revenue per month', text_auto=True, 
labels={'Total_Bill': 'Total Revenue'}).show()
```

- Conclusions: 
    - Mondays have the highest revenue per weekdays
    - Fridays have the highest revenue per weekends
    - June has the highest revenue

#### Which store has highest revenue?

```python
df['store_location'].unique()
```

```python
df_revperstore = df.groupby('store_location')['Total_Bill'].sum().reset_index().sort_values(by= 'Total_Bill', ascending= False)
df_revperstore
```

```python
px.pie(data_frame=df_revperstore, names= 'store_location', values= 'Total_Bill', title= 'Revenue per store location', hole=0.5).show()
```

- Hell's Kitchen has the highest revenue (all are approximately equal)

#### Which products sell most frequently?

```python
df_popularProduct = df['product_category'].value_counts().reset_index() 
df_popularProduct
```

```python
# top 5 product category
px.histogram(data_frame=df_popularProduct.head(5), x='product_category', y='count', title= 'Top 5 product category', text_auto=True, labels={'product_category':'Products'}).show()
```

#### Most popular product category

```python
px.pie(data_frame=df_popularProduct, names= 'product_category', values= 'count', title= 'Most popular product category', hole=0.5).show()
```

- Coffee is the most popular product

#### Product category and type distribution

```python
df_chart = df.groupby(['product_category'])['product_type'].value_counts().reset_index()
df_chart
```

```python
px.bar(data_frame=df_chart, y='product_category', x='count', color= 'product_type', 
       title= 'Product category and type', text_auto=True,
       labels={'product_category':'Categories', 'count': 'Total number'}).show()
```

#### Revenue over time trends

```python
df_daterev =  df.groupby(df['transaction_date'].dt.date)['Total_Bill'].sum().reset_index()
df_daterev
```

```python
px.line(data_frame=df_daterev, x='transaction_date', y='Total_Bill', title= 'Revenue over time').show()
```

#### Trends of revenue per hour

```python
df_salesperhour = df.groupby(df['Hour'])['Total_Bill'].sum().reset_index()
df_salesperhour
```

```python
px.line(data_frame=df_salesperhour, x='Hour', y='Total_Bill', title='Trends of revenue per hour',labels={'Hour':'Clock Hour' , 'Total_Bill': 'Revenue per Hour'}).show()
```

#### Total Revenue by Day and Hour

```python
sales_pivot = df.pivot_table(values='Total_Bill', index='Day Name', columns='Hour', aggfunc='sum')

#Order days correctly
day_order = [
   'Monday', 'Tuesday', 'Wednesday',
   'Thursday', 'Friday', 'Saturday', 'Sunday'
]

sales_pivot = sales_pivot.reindex(day_order)

# Plot heatmap
px.imshow(sales_pivot, text_auto=True, title='Total Revenue by Day and Hour').show()
```

- Range from 8:00 am till 10:00 am is the peak

# Feature Engineering

#### Average bill per time of day ('Morning', 'Afternoon', 'Evening', 'Night')

```python
df['Day_time'] = df['Hour'].apply(lambda x: 'Morning' if  x < 12 else ('Afternoon' if  x < 16 else ('Evening' if  x < 21 else 'Night')))
df.head(3)
```

```python
df_avgperdaytime = df.groupby('Day_time')['Total_Bill'].mean().round(2).reset_index()
df_avgperdaytime
```

```python
px.pie(data_frame=df_avgperdaytime, names= 'Day_time', values= 'Total_Bill', title= 'Average bill per time of day').show()
```

#### Product category per time of day

```python
df_popcoffee = df.groupby(['Day_time', 'product_category'])['product_category'].count().reset_index(name='count').sort_values(['Day_time', 'count'], ascending=[True, False])
df_popcoffee
```

```python
px.bar(data_frame=df_popcoffee, x='Day_time', y='count', color='product_category', barmode='group', title='Product category per time of day', labels={'Day_time': 'Time of Day', 'count': 'Number of Sales', 'product_category': 'Product Category'}).show()
```

#### Revenue for each day type [Weekends, Weekdays]

```python
df['Day_Type'] = 'Weekday'
df.loc[df['Day of Week'].isin([4,5]), 'Day_Type'] = 'Weekend'
df.head()
```

```python
df_revdaytype = df.groupby('Day_Type')['Total_Bill'].sum().reset_index()
df_revdaytype
```

```python
px.bar(data_frame=df_revdaytype, x='Day_Type', y='Total_Bill', title='Revenue for each day type', text_auto= True ,labels={'Day_Type': 'Weekdays Vs Weekends', 'Total_Bill': 'Total Revenue'}).show()
```

# Conclusion

- The coffee shop generated a total revenue of $698,812 across 149,116 transactions, with June being the peak month ($166,486), Hell's Kitchen leading store performance ($236,511), and the 8–10 AM window consistently driving the highest daily sales across all locations and weekdays.


- Coffee dominated product demand (58,416 transactions), and since revenue was evenly spread across weekdays with Monday and Friday slightly ahead, the business should prioritize morning-hour promotions of coffee and top-selling bakery items to capitalize on proven peak-demand patterns and further grow profitability.
